### **Mapping Your Workflow for Cloud Integration**
*Now, let’s take a closer look at your existing research workflow to identify which parts could be enhanced through cloud computing. Start by breaking down your workflow into distinct steps, then assign each step to a separate box. Before filling in any content, decide how many steps (and thus boxes) you'll need. For each step, also indicate the intended host environment—whether it runs locally or in the cloud—by writing “Local” or “Cloud” in the box beneath it.*

```python
from IPython.display import display, HTML

BuildingBox = """
<style>
  #diagram-wrapper {
    display: flex;
    flex-direction: column;
    align-items: center;
    font-family: sans-serif;
  }
  #diagram-title {
    font-size: 2em;
    margin-bottom: 10px;
    font-weight: bold;
  }
  #controls {
    margin-bottom: 15px;
  }
  #diagram-rows {
    display: flex;
    flex-direction: column;
    gap: 10px;
  }
  .diagram-container {
    display: flex;
    align-items: center;
  }
  .diagram-container.bottom .diagram-step:after {
    content: '';
  }
  .diagram-label {
    border: 2px solid #333;
    padding: 10px;
    margin: 0 5px;
    min-width: 80px;
    text-align: center;
    font-weight: bold;
    background: #f0f0f0;
  }
  .diagram-label:after {
    content: '';
  }
  .diagram-step {
    border: 2px solid #333;
    padding: 10px;
    margin: 0 5px;
    position: relative;
    min-width: 100px;
    text-align: center;
  }
  .diagram-step:after {
    content: '→';
    position: absolute;
    right: -20px;
    top: 50%;
    transform: translateY(-50%);
    font-size: 20px;
    color: #333;
  }
  .diagram-step:last-child:after {
    content: '';
  }
  input.step-input {
    width: 100%;
    border: none;
    outline: none;
    text-align: center;
    font-size: 14px;
  }
</style>

<div id="diagram-wrapper">
  <h2 id="diagram-title">🔹 Geophysical WorkFlow</h2>
  <div id="controls">
    Number of Steps:
    <input type="number" id="step-count" value="3" min="1" max="10" style="width: 50px;">
    <button id="update-diagram">Update Diagram</button>
  </div>
  <div id="diagram-rows">
    <div class="diagram-container top" id="diagram-top"></div>
    <div class="diagram-container bottom" id="diagram-bottom"></div>
  </div>
</div>

<script>
  function makeLabel(text) {
    const lbl = document.createElement('div');
    lbl.className = 'diagram-label';
    lbl.innerText = text;
    return lbl;
  }

  function makeStepDiv() {
    const div = document.createElement('div');
    div.className = 'diagram-step';
    const input = document.createElement('input');
    input.type = 'text';
    input.className = 'step-input';
    input.placeholder = 'Up to 5 words';
    input.addEventListener('input', () => {
      const words = input.value.trim().split(/\\s+/);
      if (words.length > 5) {
        input.value = words.slice(0, 5).join(' ');
      }
    });
    div.appendChild(input);
    return div;
  }

  function buildDiagram() {
    const top = document.getElementById('diagram-top');
    const bot = document.getElementById('diagram-bottom');
    top.innerHTML = '';
    bot.innerHTML = '';

    // Prepend labels
    top.appendChild(makeLabel('Steps'));
    bot.appendChild(makeLabel('Host'));

    const count = Math.min(Math.max(parseInt(document.getElementById('step-count').value, 10), 1), 10);
    for (let i = 0; i < count; i++) {
      top.appendChild(makeStepDiv());
      bot.appendChild(makeStepDiv());
    }
  }

  document.getElementById('update-diagram').addEventListener('click', buildDiagram);
  buildDiagram();
</script>
"""

display(HTML(BuildingBox))
```
