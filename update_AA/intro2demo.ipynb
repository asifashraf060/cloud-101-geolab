```python
import sys
import subprocess

# Install Python dependencies if missing
try:
    import ipywidgets as widgets
except ImportError:
    subprocess.check_call([sys.executable, "-m", "pip", "install", "ipywidgets"])
    import ipywidgets as widgets

try:
    import graphviz
except ImportError:
    subprocess.check_call([sys.executable, "-m", "pip", "install", "graphviz"])
    import graphviz

from IPython.display import display, clear_output

# Slider to select number of boxes (steps)
steps_slider = widgets.IntSlider(value=3, min=1, max=10, step=1, description='Steps:')

# Container for text inputs
box_container = widgets.VBox()

# Output area for the diagram
graph_output = widgets.Output()

def update_graph(*args):
    """Generate and display the diagram based on current text inputs."""
    with graph_output:
        clear_output()
        dot = graphviz.Digraph(format='svg')
        dot.attr(rankdir='LR')            # Left-to-right orientation
        dot.attr('node', shape='box')     # Boxes for nodes
        
        # Add nodes
        for i, text_widget in enumerate(box_container.children):
            label = text_widget.value.strip()
            dot.node(f's{i}', label or f'Step {i+1}')
        
        # Connect nodes with arrows
        n = steps_slider.value
        for i in range(n - 1):
            dot.edge(f's{i}', f's{i+1}')
        
        display(dot)

def on_text_change(change):
    """Enforce a maximum of 5 words per box, then update the diagram."""
    words = change['new'].strip().split()
    if len(words) > 5:
        # Truncate to first 5 words
        change['owner'].value = ' '.join(words[:5])
    update_graph()

def on_steps_change(change):
    """Rebuild text input widgets when the number of steps changes."""
    n = change['new']
    text_widgets = []
    for i in range(n):
        tw = widgets.Text(
            value='',
            placeholder='Up to 5 words',
            description=f'Step {i+1}:',
            layout=widgets.Layout(width='50%')
        )
        tw.observe(on_text_change, names='value')
        text_widgets.append(tw)
    box_container.children = text_widgets
    update_graph()

# Wire up the slider and initialize
steps_slider.observe(on_steps_change, names='value')
on_steps_change({'new': steps_slider.value})

# Display the interactive controls
display(steps_slider, box_container, graph_output)
```
